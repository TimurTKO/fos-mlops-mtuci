# Исследовательский ML-прототип

Цель: обучить простую модель бинарной классификации и сохранить метрики. В ЛР1 этот прототип преобразуется в структурированный Python-проект.

```
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```
train = pd.read_csv("../data/train.csv")
features = [c for c in train.columns if c.startswith("feature_")]
X, y = train[features], train["target"]
train.head()
```

```
model = Pipeline([("scaler", StandardScaler()), ("classifier", LogisticRegression(max_iter=1000, random_state=42))])
model.fit(X, y)
f1_score(y, model.predict(X))
```
